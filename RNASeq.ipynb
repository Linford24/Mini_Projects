```python
for num in range(2, 5):
  # Align a single trimmed sample
  !STAR \
     --genomeDir '/content/workspace/genomeIndex' \
     --runThreadN 2 \
     --readFilesIn '/content/workspace/trimmed/Arabidopsis_sample{num}_qc.fq' \
     --outFileNamePrefix '/content/workspace/mapped/Arabidopsis_sample{num}_qc' \
     --outSAMtype BAM SortedByCoordinate \
     --outSAMunmapped None \
     --outFilterMismatchNmax 3 \
     --outFilterMultimapNmax 1 \
     --outSAMattributes All
```

```python
!rm -rf /content/workspace/trimmed/*
```

```python
for num in range(2, 5):
    input_file = f"/content/Arabidopsis_sample{num}.fq.gz"
    # Added .gz to the end here!
    output_file = f"/content/workspace/trimmed/Arabidopsis_sample{num}_qc.fq.gz"

    print(f"Trimming sample {num}...")

    !TrimmomaticSE \
      -phred33 \
      -threads 1 \
      {input_file} \
      {output_file} \
      ILLUMINACLIP:adapters.fasta:2:30:10 \
      LEADING:3 \
      TRAILING:3 \
      SLIDINGWINDOW:4:15 \
      MINLEN:25
```

```python
import os
os.makedirs("/content/workspace/mapped", exist_ok=True)

for num in range(2, 5):
    # Pointing to the new .gz files
    input_file = f"/content/workspace/trimmed/Arabidopsis_sample{num}_qc.fq.gz"
    output_prefix = f"/content/workspace/mapped/Arabidopsis_sample{num}_"

    print(f"Aligning sample {num} with STAR...")

    !STAR \
      --genomeDir '/content/workspace/genomeIndex' \
      --runThreadN 2 \
      --readFilesIn {input_file} \
      --readFilesCommand zcat \
      --outFileNamePrefix {output_prefix} \
      --outSAMtype BAM SortedByCoordinate \
      --outSAMunmapped None \
      --outFilterMismatchNmax 3 \
      --outFilterMultimapNmax 1 \
      --outSAMattributes All
```
